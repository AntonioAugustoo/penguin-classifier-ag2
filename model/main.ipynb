````python
import pandas as pd

# 1. Coleta e leitura do conjunto de dados a partir do CSV no Drive
file_path = '/content/drive/MyDrive/penguin-classifier-ag2-main/model/palmerpenguins.csv'
df_raw = pd.read_csv(file_path)

display(df_raw.head())
````

```markdown
### 2. Mapeamento Categórico
Vamos converter as variáveis categóricas (Espécie, Ilha e Sexo) em números inteiros para que o modelo possa processá-las.
```

````python
import pandas as pd

# Criando cópia para processamento
df_processed = df_raw.copy()

# Mapeamento de Espécie
species_map = {val: i for i, val in enumerate(df_processed['species'].unique())}
# Mapeamento de Ilha
island_map = {val: i for i, val in enumerate(df_processed['island'].unique())}
# Mapeamento de Sexo (ajustado para tratar NaNs e strings)
sex_map = {'MALE': 0, 'FEMALE': 1}

df_processed['species'] = df_processed['species'].map(species_map)
df_processed['island'] = df_processed['island'].map(island_map)
df_processed['sex'] = df_processed['sex'].map(sex_map)

print("Mapeamentos aplicados:")
print(f"Espécies: {species_map}")
print(f"Ilhas: {island_map}")
print(f"Sexo: {sex_map}")
````

```markdown
### 3. Limpeza e Organização
Nesta etapa, removeremos valores nulos e organizaremos as colunas.
```

````python
# Tratamento de valores nulos
df_cleaned = df_processed.dropna().copy()

# Garantir que o sexo seja inteiro
df_cleaned['sex'] = df_cleaned['sex'].astype(int)

# Reordenação de colunas baseada no seu CSV específico
# Colunas disponíveis: species, island, culmen_length_mm, culmen_depth_mm, flipper_length_mm, body_mass_g, sex
cols = ['island', 'culmen_length_mm', 'culmen_depth_mm', 'flipper_length_mm', 'body_mass_g', 'sex', 'species']
df_final = df_cleaned[cols]

print(f"Tamanho original: {len(df_raw)}")
print(f"Tamanho após limpeza: {len(df_final)}")
display(df_final.head())
````

````python
# Salvando o dataset final aprimorado
output_path = '/content/drive/MyDrive/penguin-classifier-ag2-main/model/palmerpenguins_final.csv'
df_final.to_csv(output_path, index=False)

print(f"Dataset salvo com sucesso em: {output_path}")
````
